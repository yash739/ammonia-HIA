```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits

# Load the FITS file
with fits.open("/home/yasho379/ammonia_hfs/Magritte/testing/simulated_nh3_spectrum_oneone.fits") as hdul:
    data = hdul[0].data
    header = hdul[0].header

# Extract relevant header values
crval1 = header['CRVAL1']   # Reference velocity
cdelt1 = header['CDELT1']   # Velocity increment per pixel
crpix1 = header['CRPIX1']   # Reference pixel
naxis1 = header['NAXIS1']   # Number of pixels

# Construct the velocity array
velocity = crval1 + (np.arange(naxis1) + 1 - crpix1) * cdelt1

# Normalize by maximum value
data_normalized = data / np.max(data)

print(f"Original data range: {data.min():.3e} to {data.max():.3e}")
print(f"Normalized data range: {data_normalized.min():.6f} to {data_normalized.max():.6f}")

# Save as simple ASCII file (no header)
# Two columns: velocity, normalized intensity
ascii_data = np.column_stack([velocity, data_normalized])
np.savetxt("nh3_11_spectrum.txt", ascii_data, fmt="%.6f")

print(f"\nSaved {len(velocity)} data points to nh3_11_spectrum.txt")
print(f"Format: velocity (km/s), normalized intensity")
print(f"Velocity range: {velocity.min():.2f} to {velocity.max():.2f} km/s")

# Plot Intensity vs. Velocity
plt.figure(figsize=(10, 5))
plt.plot(velocity, data_normalized, lw=1)
plt.xlabel('Velocity (km/s)')
plt.ylabel('Normalized Intensity')
plt.title('NH₃(1,1) Spectrum (Normalized)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pyspeckit
from pyspeckit import Spectrum
from pyspeckit.wrappers.fitnh3 import fitnh3tkin

# === STEP 1: Load and Normalize Spectra ===
# Replace with your actual file paths
file_oneone = '/home/yasho379/ammonia_hfs/Magritte/testing/simulated_nh3_spectrum_oneone.fits'
file_twotwo = '/home/yasho379/ammonia_hfs/Magritte/testing/simulated_nh3_spectrum_twotwo.fits'

# Load spectra
spec_11 = Spectrum(file_oneone)
spec_22 = Spectrum(file_twotwo)

# Normalize data to max = 1 (important!)
spec_11.data /= spec_11.data.max()
spec_22.data /= spec_22.data.max()

# Set unit label
spec_11.unit = 'normalized intensity'
spec_22.unit = 'normalized intensity'

# Convert x-axis to velocity (km/s)
spec_11.xarr.convert_to_unit('km/s')
spec_22.xarr.convert_to_unit('km/s')

# Optionally set rest frequency to NH3(1,1)
spec_11.xarr.refX = 23.694495e9  # Hz
spec_11.xarr.convert_to_unit('km/s')
spec_22.xarr.refX = 23.7226333e9  # Hz for NH3(2,2)
spec_22.xarr.convert_to_unit('km/s')

# Create dictionary of loaded spectra
spectra_dict = {
    'oneone': spec_11,
    'twotwo': spec_22
}

# === STEP 2: Fit the NH₃ spectra ===
spdict, result = fitnh3tkin(
    spectra_dict,
    guessline='oneone',
    crop=False,
    rebase=True,
    smooth=True,
    doplot=True,
    verbose=True,
    npeaks=1,
    fittype='ammonia',

    # Improved initial guesses
    tkin=20,            # Kinetic temperature [K]
    tex=5,              # Excitation temperature [K]
    column=14.5,        # log10(N) [cm^-2]
    fortho=0.5          # Ortho-to-para ratio
)

# === STEP 3: Print fitted parameters ===
print("\nFitted parameters for NH₃(1,1):")
for p in spdict['oneone'].specfit.parinfo:
    print(f"{p.name}: {p.value:.4f} ± {p.error:.4f}")

print("\nFitted parameters for NH₃(2,2):")
for p in spdict['twotwo'].specfit.parinfo:
    print(f"{p.name}: {p.value:.4f} ± {p.error:.4f}")
```

```python
import numpy as np
import pyspeckit
from astropy import units as u
from pyspeckit.spectrum.models import ammonia_constants, ammonia, ammonia_hf
from pyspeckit.spectrum.models.ammonia_constants import freq_dict
from pyspeckit.spectrum.units import SpectroscopicAxis, SpectroscopicAxes

# Step 1. Generate a synthetic spectrum.  Already have a real spectrum?  Skip
# to step 2!
# Generate a synthetic spectrum based off of 3 NH3 lines
# Note that they are converted to GHz first
xarr11 = SpectroscopicAxis(np.linspace(-30, 30, 100)*u.km/u.s,
                           velocity_convention='radio',
                           refX=freq_dict['oneone']).as_unit(u.GHz)
xarr22 = SpectroscopicAxis(np.linspace(-40, 40, 100)*u.km/u.s,
                           velocity_convention='radio',
                           refX=freq_dict['twotwo']).as_unit(u.GHz)
xarr33 = SpectroscopicAxis(np.linspace(-50, 50, 100)*u.km/u.s,
                           velocity_convention='radio',
                           refX=freq_dict['threethree']).as_unit(u.GHz)
# Merge the three X-axes into a single axis
xarr = SpectroscopicAxes([xarr11,xarr22,xarr33])

# Compute a synthetic model that is made of two temperature components with
# identical velocities
synthspec = (ammonia.ammonia(xarr, trot=20, ntot=15, fortho=0.5, xoff_v=0.0,
                             width=1.0) +
             ammonia.ammonia(xarr, trot=50, ntot=14, fortho=0.5, xoff_v=0.0,
                             width=1.0))

# Create the Spectrum object
spectrum = pyspeckit.Spectrum(xarr=xarr, data=synthspec, header={})


# Step 2.  You have a spectrum.
# plot it
spectrum.plotter()

# Use the multi-tex/multi-tau model generator to build up a model function
# You can use any set of oneone, twotwo, ..., eighteight (no 9-9 or higher)
# This sets the number of parameters to be fit: 2+2*(n_transitions)
fitter = ammonia_hf.nh3_vtau_multimodel_generator(['oneone', 'twotwo',
                                                   'threethree'])

# Register the fitter - i.e., tell pyspeckit where it is and how to use it
spectrum.specfit.Registry.add_fitter('nh3_vtau_123', fitter, fitter.npars)
# These are the parameter names, approximately:
# parnames=['center','width','Tex11','tau11','Tex22','tau22','Tex33','tau33'],

# Need to give some input guesses.  We start with something wrong-ish: -5 km/s,
# 1.2 km/s width, and 15 K + 0.5 tau for all 3 lines
guesses = [-5, 1.2, 15, 0.5, 15, 0.5, 15, 0.5,]

# Plot up the guessed model
spectrum.plotter.axis.plot(spectrum.xarr,
                           fitter.n_modelfunc(guesses)(spectrum.xarr), 'b')

# Run the fit!
spectrum.specfit(fittype='nh3_vtau_123', guesses=guesses)

# display the correct and fitted answers
print("Low column version:")
def printthings(ammonia=ammonia.ammonia, xarr=xarr):
    print("Real optical depths of component 1: ",[ammonia(xarr, trot=20,
                                                          ntot=15, fortho=0.5,
                                                          xoff_v=0.0,
                                                          width=1.0,
                                                          return_tau=True)[x]
                                                  for x in ['oneone', 'twotwo',
                                                            'threethree']])
    print("Real optical depths of component 2: ",[ammonia(xarr, trot=50,
                                                          ntot=14, fortho=0.5,
                                                          xoff_v=0.0,
                                                          width=1.0,
                                                          return_tau=True)[x]
                                                  for x in ['oneone', 'twotwo',
                                                            'threethree']])
printthings()

print("Fitted parameters: ",spectrum.specfit.parinfo)

# It works, but the covariances between tex and tau are large.
# So, another example with higher tau (and therefore... less degenerate?)

synthspec = (ammonia.ammonia(xarr, trot=20, ntot=16, fortho=0.5, xoff_v=0.0,
                             width=1.0) +
             ammonia.ammonia(xarr, trot=50, ntot=15, fortho=0.5, xoff_v=0.0,
                             width=1.0))
spectrum2 = pyspeckit.Spectrum(xarr=xarr, data=synthspec, header={})
spectrum2.plotter()
spectrum2.specfit.Registry.add_fitter('nh3_vtau_123', fitter, fitter.npars)
spectrum2.specfit(fittype='nh3_vtau_123', guesses=guesses)

# We can also examine what tau really should have been... kinda.
print("High column version:")
def printthings(ammonia=ammonia.ammonia, xarr=xarr):
    print("Real optical depths of component 1: ",[ammonia(xarr, trot=20,
                                                          ntot=16, fortho=0.5,
                                                          xoff_v=0.0,
                                                          width=1.0,
                                                          return_tau=True)[x]
                                                  for x in ['oneone', 'twotwo',
                                                            'threethree']])
    print("Real optical depths of component 2: ",[ammonia(xarr, trot=50,
                                                          ntot=15, fortho=0.5,
                                                          xoff_v=0.0,
                                                          width=1.0,
                                                          return_tau=True)[x]
                                                  for x in ['oneone', 'twotwo',
                                                            'threethree']])
printthings()
print("Fitted parameters: ",spectrum2.specfit.parinfo)
```

```python
import pyspeckit
import numpy as np
import matplotlib.pyplot as plt

# Load spectra first to normalize
spec_11 = pyspeckit.Spectrum('/home/yasho379/ammonia_hfs/Magritte/testing/simulated_nh3_spectrum_oneone.fits')
spec_22 = pyspeckit.Spectrum('/home/yasho379/ammonia_hfs/Magritte/testing/simulated_nh3_spectrum_twotwo.fits')

print("Original data ranges:")
print(f"  NH₃(1,1): {np.min(spec_11.data):.3f} to {np.max(spec_11.data):.3f}")
print(f"  NH₃(2,2): {np.min(spec_22.data):.3f} to {np.max(spec_22.data):.3f}")

# Method: Normalize both to have baseline=0 and NH3(1,1) peak=1
# This should make them consistent

# For NH₃(1,1): subtract min, divide by (max-min)
spec_11_min = np.min(spec_11.data)
spec_11_max = np.max(spec_11.data)
spec_11.data = (spec_11.data - spec_11_min) / (spec_11_max - spec_11_min)

# For NH₃(2,2): subtract min, divide by (max-min) 
spec_22_min = np.min(spec_22.data)
spec_22_max = np.max(spec_22.data)
spec_22.data = (spec_22.data - spec_22_min) / (spec_22_max - spec_22_min)

print(f"After individual normalization:")
print(f"  NH₃(1,1): {np.min(spec_11.data):.3f} to {np.max(spec_11.data):.3f}")
print(f"  NH₃(2,2): {np.min(spec_22.data):.3f} to {np.max(spec_22.data):.3f}")

# Plot the normalized spectra
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

# NH₃(1,1) plot
axes[0].plot(spec_11.xarr.value, spec_11.data, 'b-', linewidth=1.5)
axes[0].set_ylabel('Normalized Intensity')
axes[0].set_title('NH₃(1,1) - After Normalization (0 to 1)')
axes[0].grid(True, alpha=0.3)
axes[0].set_ylim(-0.1, 1.1)

# NH₃(2,2) plot
axes[1].plot(spec_22.xarr.value, spec_22.data, 'r-', linewidth=1.5)
axes[1].set_xlabel('Velocity (km/s)')
axes[1].set_ylabel('Normalized Intensity')
axes[1].set_title('NH₃(2,2) - After Normalization (0 to 1)')
axes[1].grid(True, alpha=0.3)
axes[1].set_ylim(-0.1, 1.1)

plt.tight_layout()
plt.show()

# Create dictionary with normalized spectra
spectra_dict = {
    'oneone': spec_11,
    'twotwo': spec_22
}

# Fit NH₃
spdict, spectra = pyspeckit.wrappers.fitnh3.fitnh3tkin(
    spectra_dict,
    tkin=12,
    tex=8, 
    column=13.5,
    fortho=0.3,
    verbose=True
)

# Print results
print("\nFitted parameters:")
spec = spdict['oneone']
if hasattr(spec.specfit, 'parinfo'):
    try:
        for i, param in enumerate(spec.specfit.parinfo):
            if hasattr(param, 'name'):
                print(f"{param.name}: {param.value:.3f}")
            else:
                # Parameter names in order for NH3 fitting
                param_names = ['trot', 'tex', 'ntot', 'width', 'xoff_v', 'fortho']
                if i < len(param_names):
                    print(f"{param_names[i]}: {param.value:.3f}")
                else:
                    print(f"param_{i}: {param.value:.3f}")
    except:
        print("Parameters fitted successfully - check the iteration output above")
else:
    print("No fit parameters found")
```
